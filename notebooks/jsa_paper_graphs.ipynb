```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

plt.rcParams.update({
    "text.color": "black",
    "axes.labelcolor": "black",
    "xtick.color": "black",
    "ytick.color": "black",
    "legend.labelcolor": "black"
})

# === Component names (OVERHEAD REMOVED) ===
components = ['Tokenization', 'MSA', 'MLP', 'Layer Norm', 'Residual', 'ViT I/O']

# === Timing data (LAST COLUMN REMOVED) ===
base_times   = [0.683, 22.008, 45.108, 1.368, 0.696, 0.572]
drop30_times = [0.683, 20.412, 40.020, 1.176, 0.696, 0.572]
drop50_times = [0.683, 15.816, 32.196, 0.984, 0.696, 0.572]

data = np.array([base_times, drop30_times, drop50_times])

models = [
    'GateAttn-ViT-Small (0)',
    'GateAttn-ViT-Small (10)',
    'GateAttn-ViT-Small (20)'
]

# === Transform function for Y-axis ===
def transform(y):
    if y <= 1.6:
        return (y / 1.6) * 0.5
    else:
        return 0.5 + (y - 1.6) / (50 - 1.6) * 0.5

def apply_transform(values):
    return [transform(v) for v in values]

transformed_data = np.array([apply_transform(times) for times in data])

# === Plot parameters ===
n_components = len(components)
n_models = len(models)

group_spacing = 0.8
bar_width = 0.2
x = np.arange(n_components) * group_spacing
offsets = np.linspace(-bar_width, bar_width, n_models)

colors = ["#4C72B0", "#DD8452", "#55A868"]

# === Plot ===
fig, ax = plt.subplots(figsize=(6.0, 3.0))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

for i, (label, color) in enumerate(zip(models, colors)):
    ax.bar(
        x + offsets[i],
        transformed_data[i],
        width=bar_width * 0.9,
        label=label,
        color=color,
        edgecolor='black',
        linewidth=0.0,
        zorder=3
    )

# === X-axis ===
ax.set_xticks(x)
ax.set_xticklabels(components, fontsize=12, rotation=30, ha='center')

# === Y-axis ===
yticks_raw = list(np.arange(0.1, 1.6, 0.25)) + [4, 12, 20, 30, 40, 50]
yticks_trans = [transform(y) for y in yticks_raw]
ax.set_yticks(yticks_trans)
ax.set_yticklabels([f"{y:.1f}" for y in yticks_raw], fontsize=12)
ax.set_ylim(0, 1)
ax.set_ylabel("Time (milliseconds)", fontsize=12)

# Grid
ax.yaxis.grid(True, linestyle="--", linewidth=0.5, alpha=0.1, color="gray", zorder=0)

# Spines
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.spines['left'].set_color('black')
ax.spines['bottom'].set_color('black')

# Legend
legend = fig.legend(
    *ax.get_legend_handles_labels(),
    fontsize=10,
    loc='upper center',
    bbox_to_anchor=(0.71, 0.88),
    frameon=True
)

plt.subplots_adjust(top=0.78)
plt.savefig("time_small_no_overhead.svg", bbox_inches="tight", format="svg")
plt.show()
```
