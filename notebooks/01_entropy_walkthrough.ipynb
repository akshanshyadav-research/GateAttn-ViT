# Entropy-gated patch selection
This walkthrough uses a deterministic synthetic image, without datasets or pretrained weights. See the README for evaluation and attention visualization commands.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from gateattn.entropy import patch_entropy, select_mask, baseline_mask
from gateattn.visualize import overlay_mask
rng = np.random.default_rng(2026)
image = np.full((224, 224, 3), 235, dtype=np.uint8)
image[48:176, 64:160] = rng.integers(0, 256, (128, 96, 3), dtype=np.uint8)
entropy = patch_entropy(image)
keep = select_mask(entropy, drop_rate=0.5, policy="stable")
print(f"Retained {keep.sum()} of {len(keep)} patches")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(10, 3))
axes[0].imshow(image); axes[0].set_title("Synthetic input")
axes[1].imshow(entropy.reshape(14, 14), cmap="magma"); axes[1].set_title("Entropy (bits)")
axes[2].imshow(overlay_mask(image, keep)); axes[2].set_title("50% retained")
for ax in axes: ax.axis("off")
plt.tight_layout()
plt.show()
```
